````python
def is_prime2(x):
    """Check if x is prime using trial division."""
    if x < 2:
        return False
    if x == 2:
        return True
    if x % 2 == 0:
        return False
    
    # Only check odd divisors up to sqrt(x)
    i = 3
    while i * i <= x:
        if x % i == 0:
            return False
        i += 2
    return True
````

````python
is_prime2(10**(10**8)+1)
````

````python
def is_prime_with_gen(x):
    """Check if x is prime using the prime generator."""
    if x < 2:
        return False
    
    g = gen_primes()
    for prime in g:
        if prime * prime > x:
            return True  # No divisors found up to sqrt(x)
        if x % prime == 0:
            return x == prime  # True only if x itself is the prime
````

````python
is_prime_with_gen(10**10**8+1)
````

````python
def is_divisible(lst, x):
    for i, num in enumerate(lst):
        lst[i] = 'yes' if num%x == 0 else 'no'
    return lst
````

````python
def sort_words(s, letter):
    return list(filter(lambda x : x[-1] == letter, s.split(' ')))
````

````python
sort_words('I am in love with python', 'n')
````

````python
## Index of coincidence
````

````python
def coincidence_index(s, subs):
    lst = []
    for i, word in enumerate(s.split(' ')):
        if subs in word:
            lst.append(i)
    return lst
````

````python
def coincidence_index(s, subs):
    return [i for i, x in enumerate(list(map(lambda x: subs in x, s.split(' ')))) if x]
````

````python
def tax_calc(lst):
    return reduce(lambda x: x[0]*x[1], lst), 0)
````
